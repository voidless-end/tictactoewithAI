# 🤖 AI Battle Arena - Tic-Tac-Toe Tournament Demo

This notebook demonstrates the AI Battle Arena system where different AI strategies compete against each other on a variable-dimension tic-tac-toe board. You'll see:

- **Heat maps** showing where each AI prefers to make moves
- **Win rate comparisons** between different AI strategies
- **Comprehensive statistics** including game durations and move analysis
- **Strategic insights** from visual analytics of AI behavior

Let's run a tournament and analyze the results!

````python
# Import all required modules
from ai_battle_arena import AIBattleArena, TournamentStats
from ai_statistics import BattleStatistics
from ai_strategies import AI_TYPES
import numpy as np
import matplotlib.pyplot as plt

# Configure matplotlib for notebook
%matplotlib inline
plt.rcParams['figure.figsize'] = (12, 8)

print("✅ All imports successful!")
print(f"Available AI strategies: {', '.join(AI_TYPES)}")
````

## 1. Run a Tournament

Let's run a 30-game tournament between Minimax (optimal play) and Random (baseline strategy) on a 3x3 board.

````python
# Initialize the battle arena
arena = AIBattleArena(board_size=3, verbose=False)

# Run tournament: Minimax vs Random (30 games)
print("🎮 Running 30-game tournament: Minimax vs Random...")
stats = arena.run_tournament('minimax', 'random', num_games=30)

# Display tournament results
print(f"\n📊 Tournament Results:")
print(f"  Minimax wins: {stats.ai1_wins} ({stats.ai1_win_rate:.1f}%)")
print(f"  Random wins:  {stats.ai2_wins} ({stats.ai2_win_rate:.1f}%)")
print(f"  Draws:        {stats.draws} ({stats.draw_rate:.1f}%)")
print(f"  Average game duration: {stats.avg_game_duration:.4f}s")
print(f"  Average moves per game: {stats.avg_moves_per_game:.1f}")
````

## 2. Move Heatmaps

Heatmaps show which board positions each AI prefers to play. This reveals strategic preferences:
- **Minimax**: Should show balanced positions, focusing on optimal play theory
- **Random**: Should show uniform distribution across all positions

````python
# Generate move heatmaps
bs = BattleStatistics()
fig, path = bs.plot_move_heatmaps(stats, save=False, show=False)
plt.show()

print("\n📍 Heatmap interpretation:")
print("  - Darker colors = More frequently played positions")
print("  - Minimax tends to favor center and corners (classic tic-tac-toe strategy)")
print("  - Random shows more uniform distribution (no preference)")
````

## 3. Winning Positions Analysis

These heatmaps show where each AI places their **winning moves** - the actual moves that clinch victory. This reveals tactical strengths.

````python
# Generate winning positions heatmaps
fig, path = bs.plot_winning_positions(stats, save=False, show=False)
plt.show()

print("\n🎯 Winning position analysis:")
print("  - Minimax's winning moves should be well-distributed (multiple paths to victory)")
print("  - Random's winning positions will be sparse (few games won)")
print("  - Concentration in corners/edges indicates defensive playing patterns")
````

## 4. First Move Distribution

Analyze opening move preferences across all games.

````python
# Generate first move distribution heatmap
fig, path = bs.plot_first_moves(stats, save=False, show=False)
plt.show()

print("\n🎯 First move insights:")
print("  - Center is typically the strongest opening in tic-tac-toe")
print("  - Corners are secondary strong positions")
print("  - Edge positions are generally weaker openings")
````

## 5. Win Rate Comparison

Direct comparison of win rates showing which AI is stronger and consistency across multiple metrics.

````python
# Generate win rate comparison
fig, path = bs.plot_win_rates(stats, save=False, show=False)
plt.show()

print("\n📈 Key takeaways:")
print(f"  - Minimax win rate: {stats.ai1_win_rate:.1f}% (deterministic optimal strategy)")
print(f"  - Random win rate: {stats.ai2_win_rate:.1f}% (baseline/random play)")
print(f"  - Draw rate: {stats.draw_rate:.1f}% (reflects tic-tac-toe dynamics on 3x3 board)")
````

## 6. Game Duration Analysis

Analyze how long games take and how many moves are typically played.

````python
# Generate game duration histogram
fig, path = bs.plot_game_durations(stats, save=False, show=False)
plt.show()

print("\n⏱️ Duration insights:")
print(f"  - Average duration: {stats.avg_game_duration*1000:.2f}ms")
print(f"  - Average moves: {stats.avg_moves_per_game:.1f}")
print("  - Minimax takes longer per move due to computation")
print("  - Random AI is nearly instantaneous")
````

## 7. Comprehensive Summary Dashboard

A unified dashboard showing all key metrics: move patterns, game durations, first moves, and win rates.

````python
# Generate comprehensive summary dashboard
fig, path = bs.plot_summary_dashboard(stats, save=False, show=False)
plt.show()

print("\n📊 Summary Dashboard includes:")
print("  - Win distribution pie chart")
print("  - Move heatmaps for each player")
print("  - Winning positions and first move preferences")
print("  - Game length distribution")
print("  - Win progression over time")
print("  - Comprehensive statistics summary")
````

## 8. Round-Robin Tournament

Now let's run all AI strategies against each other! This shows the complete ranking and comparative performance.

````python
# Run round-robin tournament with all AI types
print("🏆 Running round-robin tournament with all AI strategies...")
print(f"   Strategies: {', '.join(AI_TYPES)}")
print("   Mode: Each AI plays against every other AI (20 games each matchup)\n")

arena_rr = AIBattleArena(board_size=3, verbose=False)
rr_results = arena_rr.run_round_robin(ai_types=AI_TYPES, num_games=20)

# Display summary of each matchup
print("\n📊 Round-Robin Matchup Details:")
print("=" * 60)
for matchup_name, matchup_stats in rr_results.items():
    print(f"\n{matchup_name}:")
    print(f"  {matchup_stats.ai1_name}: {matchup_stats.ai1_wins} wins ({matchup_stats.ai1_win_rate:.1f}%)")
    print(f"  {matchup_stats.ai2_name}: {matchup_stats.ai2_wins} wins ({matchup_stats.ai2_win_rate:.1f}%)")
    print(f"  Draws: {matchup_stats.draws} ({matchup_stats.draw_rate:.1f}%)")
````

## 9. Round-Robin Results Visualization

Visualize the complete round-robin tournament results as a heatmap matrix.

````python
# Generate round-robin visualization
fig, path = bs.plot_round_robin_results(rr_results, save=False, show=False)
plt.show()

print("\n🎯 Round-robin matrix interpretation:")
print("  - Each cell shows win rate of Row AI against Column AI")
print("  - Green = High win rate, Red = Low win rate")
print("  - Diagonal is empty (AIs don't play themselves)")
print("  - Minimax should dominate most matchups")
print("  - Strategic and Aggressive AIs should outperform Random")
````

## 10. Compare Two Different Strategies

Test how strategies perform on a larger 4x4 board where complexity increases.

## 12. Conclusions & Next Steps

### Key Insights

1. **Strategy Effectiveness**: Minimax dominates because it plays optimally by exploring all move consequences
2. **Strategic vs Random**: Rule-based strategies (Strategic AI) outperform pure randomness but lose to optimal play
3. **Draw Rate**: On a 3x3 board, optimal play leads to frequent draws (expected for tic-tac-toe)
4. **Monte Carlo Strength**: Monte Carlo Tree Search becomes more competitive on larger boards where minimax is computationally expensive
5. **Aggressive Play**: Aggressive AI trades defense for offense, which can be effective against weaker opponents

### Expected Performance Hierarchy (3x3 board)

1. **Minimax** - Optimal play, nearly unbeatable
2. **Strategic** - Strong rule-based heuristics
3. **Monte Carlo** - Good simulation-based approach
4. **Aggressive** - Risky but can exploit weak opponents
5. **Random** - Baseline strategy

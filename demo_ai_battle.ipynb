This notebook demonstrates the AI Battle Arena system where different AI strategies compete against each other on a variable-dimension tic-tac-toe board. You'll see:

- **Heat maps** showing where each AI prefers to make moves
- **Win rate comparisons** between different AI strategies
- **Comprehensive statistics** including game durations and move analysis

Let's run a tournament and analyze the results!

```python
import sys
sys.path.insert(0, '/home/meh/Downloads/tic_tac_toe_3-master')

from ai_battle_arena import AIBattleArena, TournamentStats
from ai_statistics import BattleStatistics
from ai_strategies import AI_TYPES
import numpy as np
import matplotlib.pyplot as plt

print("✅ All imports successful!")
print(f"Available AI strategies: {', '.join(AI_TYPES)}")
```

## 1. Run a Tournament

Let's run a 30-game tournament between Minimax (optimal play) and Random (baseline strategy) on a 3x3 board.

```python
# Initialize the battle arena
arena = AIBattleArena(board_size=3, verbose=False)

# Run tournament: Minimax vs Random (30 games)
print("🎮 Running 30-game tournament: Minimax vs Random...")
stats = arena.run_tournament('minimax', 'random', num_games=30)

# Display tournament results
print(f"\n📊 Tournament Results:")
print(f"  Minimax wins: {stats.player1_wins} ({100*stats.player1_wins/30:.1f}%)")
print(f"  Random wins:  {stats.player2_wins} ({100*stats.player2_wins/30:.1f}%)")
print(f"  Draws:        {stats.draws} ({100*stats.draws/30:.1f}%)")
print(f"  Average game duration: {np.mean([g.duration for g in stats.games]):.3f}s")
print(f"  Average moves per game: {np.mean([g.num_moves for g in stats.games]):.1f}")
```

## 2. Move Heatmaps

Heatmaps show which board positions each AI prefers to play. This reveals strategic preferences:
- **Minimax**: Should show balanced positions, focusing on optimal play theory
- **Random**: Should show uniform distribution across all positions

```python
# Generate move heatmaps
bs = BattleStatistics()
fig, path = bs.plot_move_heatmaps(stats, save=False, show=True)
plt.tight_layout()
plt.show()

print("📍 Heatmap interpretation:")
print("  - Darker colors = More frequently played positions")
print("  - Minimax tends to favor center and corners (classic tic-tac-toe strategy)")
print("  - Random shows more uniform distribution (no preference)")
```

## 3. Winning Positions Analysis

These heatmaps show where each AI places their **winning moves** - the actual moves that clinch victory. This reveals tactical strengths.

```python
# Generate winning positions heatmaps
fig, path = bs.plot_winning_positions(stats, save=False, show=True)
plt.tight_layout()
plt.show()

print("🎯 Winning position analysis:")
print("  - Minimax's winning moves should be well-distributed (multiple paths to victory)")
print("  - Random's winning positions will be sparse (few games won)")
print("  - Concentration in corners/edges indicates defensive playing patterns")
```

## 4. Win Rate Comparison

Direct comparison of win rates showing which AI is stronger and consistency across multiple metrics.

```python
# Generate win rate comparison
fig, path = bs.plot_win_rates(stats, save=False, show=True)
plt.tight_layout()
plt.show()

print("📈 Key takeaways:")
print(f"  - Minimax win rate: ~90%+ (deterministic optimal strategy)")
print(f"  - Random win rate: ~0-10% (baseline/random play)")
print(f"  - Draw rate reflects tic-tac-toe dynamics on 3x3 board")
```

## 5. Comprehensive Summary Dashboard

A unified dashboard showing all key metrics: move patterns, game durations, first moves, and win rates.

```python
# Generate comprehensive summary dashboard
fig, path = bs.plot_summary_dashboard(stats, save=False, show=True)
plt.tight_layout()
plt.show()

print("📊 Summary Dashboard includes:")
print("  - Move heatmaps for each player")
print("  - Game duration histogram")
print("  - Win rate bar chart")
print("  - First move analysis showing opening preferences")
```

## 6. Round-Robin Tournament

Now let's run all AI strategies against each other! This shows the complete ranking and comparative performance.

```python
# Run round-robin tournament with all AI types
print("🏆 Running round-robin tournament with all AI strategies...")
print(f"   Strategies: {', '.join(AI_TYPES)}")
print("   Mode: Each AI plays against every other AI (10 games each matchup)\n")

arena = AIBattleArena(board_size=3, verbose=False)
rr_results = arena.run_round_robin(games_per_matchup=10)

# Display round-robin results
print("📊 Round-Robin Results Summary:")
for ai_type, record in sorted(rr_results.items(), key=lambda x: -x[1]['total_wins']):
    record = rr_results[ai_type]
    total_games = sum(record['vs'].values())
    print(f"\n{ai_type.upper():12} | Wins: {record['total_wins']:2d} | Losses: {record['total_losses']:2d} | Draws: {record['total_draws']:2d}")
    print(f"               | Win Rate: {100*record['total_wins']/(total_games+1e-8):.1f}%")
```
